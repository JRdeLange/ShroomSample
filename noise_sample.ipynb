```python
import sounddevice as sd
from pedalboard.io import AudioFile
from upath import UPath

INPUT_DEVICE = "Microphone (MPC Sample Audio), Windows WASAPI"
SAMPLE_RATE = sd.query_devices(INPUT_DEVICE, "input")["default_samplerate"]

noise_clip = sd.rec(
    int(3 * SAMPLE_RATE),
    samplerate=SAMPLE_RATE,
    channels=2,
    device=INPUT_DEVICE,
    dtype="float32",
)
sd.wait()
noise_clip = noise_clip.T  # → (channels, samples)

output_path = UPath(f"data/noise_clips/{INPUT_DEVICE}.wav")

if not output_path.parent.exists():
    output_path.parent.mkdir(parents=True)

# Save it so you don't have to re-record every session
with AudioFile(str(output_path), "w", SAMPLE_RATE, 2) as f:
    f.write(noise_clip)
```
